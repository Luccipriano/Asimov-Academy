## Introdução ao Curso - Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 5, 21)
y = x ** 2
```

```python
plt.title("Gráfico")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.plot(x, y, "green")
```

```python
plt.subplot(1, 2, 1)
plt.plot(x, y, "r-")
plt.subplot(1, 2, 2)
plt.plot(x, x, "b*-")
```

## Orientação a objetos no Matplotlib

```python
fig = plt.figure() ## Criação de uma figure
print(type(fig))

axes = fig.add_axes([0.1, 0.1, 0.8, 0.8]) # esquerda, inferior, largura, altura

axes.plot(x, y)
axes.set_xlabel("Eixo X")
axes.set_ylabel("Eixo Y")
axes.set_title("Título")
```

```python
fig = plt.figure() ## Criação de uma figure
print(type(fig))

axes = fig.add_axes([0.1, 0.1, 0.8, 0.8]) # esquerda, inferior, largura, altura
axes2 = fig.add_axes([0.2, 0.5, 0.3, 0.2])

axes.plot(x, y)
axes.set_xlabel("Eixo X")
axes.set_ylabel("Eixo Y")
axes.set_title("Título")

axes2.plot(x, x, "red")
axes2.set_xlabel("Eixo X")
axes2.set_ylabel("Eixo Y")
axes2.set_title("Título")
```
