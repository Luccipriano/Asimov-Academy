# Importações e Configurações

## Listas

```python
type([1, 2, 3])
```

```python
type([1, 2.3, True, 'Texto']) # A lista pode conter diversos tipos de dados misturados juntos
```

```python
lista_com_lista = [
    1,
    True,
    "Texto",
    [1, 2, 3]
]

# Também podemos ter listas dentro de listas
```

```python
alunos = ['Ana', 'Carlos', 'Bruno']
# Mais comumente usado pra representar dados estruturados do mundo real
```

```python
alunos[0] # A indexação de listas começa no índice 0, como o exemplo ao lado
```

```python
display(alunos[1])
alunos[2]
```

```python
# Podemos ver o tamanho de uma lista por meio da seguinte função:
len(alunos)
```

```python
# Por conta do índice começar em zero, o fim de uma lista é o tamanho dela menos 1
print(alunos)
print(alunos[len(alunos) - 1])
```

```python
# Esse acesso é facilitado por meio do uso de índices negativos
print(alunos)
print(alunos[-1]) # É como se estivesse contando do zero pra trás e dando um loop no fim da lista
print(alunos[-2])
```

```python
# Pra acessar uma lista dentro de uma lista, basta encadear os índices
lista_com_lista[-1][0]
```

```python
# Também é possível deletar valores de listas
del lista_com_lista[0]
```

```python
lista_com_lista # O valor '1' que estava no índice zero sumiu
```

## Tuplas

```python
# Tuplas são muito parecidas com listas, mas elas não aceitam atribuição e elas usam parentesis ao invés de colchetes
alunos = ('Ana', 'Maria', 'Bruno')
alunos
```

```python
alunos[0] = 'Marcio' # Não é possível fazer atribuições em valores de tuplas
```

## Sequências

```python
# Listas e Tuplas são sequências
# Strings também são sequências de caracteres
nome = 'Lucas'
nome[1] # É possível acessar os caracteres dentro de uma string da mesma forma que uma lista ou tupla
# A string é um encadeamento de caracteres por meio de ponteiros, é idêntico a uma lista
```

```python
# Sequências também podem ser vazias, geralmente ou por atribuição ou ao usar um type constructor
sequencia = ""
sequencia
```

```python
sequencia[0]
```

```python
list(sequencia)
```

```python
tuple(sequencia)
```

### Sequências verdadeiras e falsas

```python
# Sequências podem ser verdadeiras ou falsas apenas por estarem preenchidas ou não

# String vazio:
bool(str())
```

```python
bool(str("Teste"))
```

```python
bool(str(False))
```

```python
bool(False)
```

```python
bool("False")
```

```python
bool(list())
```

```python
bool(tuple())
```

```python
bool([1, 2, 3])
```

```python
bool([False])
```

```python
bool([False, True])
```

```python
print(bool((False))) # Não é uma tupla, é apenas um valor dentro de parentesis redundantes, como uma operação matemática (ou um statement)
print(bool((False,))) # Agora sim é uma tupla, devido a presença da vírgula
```

```python
bool((True))
```

```python
bool((False, True))
```

```python
bool((False, False))
```

```python
# Esse comportamente também pode ser visto em statements condicionais, que exigem um valor booleano

sequencia = [1, 2, 3]

if sequencia:
    print("A sequência possui valores")
else:
    print("A sequência não possui valores")
```
