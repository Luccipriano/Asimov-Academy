### Validando a idade do usuário

```python
idade = int(input("Qual sua idade?\nDigite aqui: "))

if idade < 18:
    print("Você é menor de idade.")
elif idade == 18:
    print("Você tem exatamente 18 anos.")
else:
    print("Você é maior de idade.")
```

## Desafio - Usuário e Senha:
- Pede por um nome de usuário e uma senha.
- Se ambos forem corretos, exibe uma mensagem de sucesso.
- Caso contrário, exibe uma mensagem de erro. A mensagem é diferente quando o usuário está incorreto, e quando a senha está incorreta
- O usuário/senha "corretos" podem ser definidos como variávies dentro do próprio código.


```python
usuario = 'Lucas'
senha = 'Lucas123'

input_usuario = input("Qual o seu usuário?\n")
input_senha = input("Qual sua senha?\n")

if usuario == input_usuario and senha == input_senha:
    print("Login efetuado com sucesso.")
elif usuario == input_usuario and senha != input_senha:
    print("Senha errada, tente novamente.")
elif usuario != input_usuario and senha == input_senha:
    print("Usuário incorreto, tente novamente.")
else:
    print("Usuário e senha incorretos, por favor tente novamente.")

# if


# input()
```

# Desafio - crie um programa que:
- Escolhe um número secreto.
- Pede por um chute do usuário.
- Indica se o usuário acertou ou não.
- Se não acertou, dá uma dica, dizendo
    - se o número é mais alto ou mais baixo.
- Repete isso até 3 vezes!

```python
print("Bem vindo ao PythonBets!\n")
time.sleep(1)
print('--------------------')
time.sleep(1)

num_rodadas = int(input("Quantas vezes você quer rodar a roleta? Escolha até 3 rodadas"))

while num_rodadas not in [1, 2, 3]:
    num_rodadas = int(input("Você escolheu um valor inválido. Escolha até 3 rodadas"))

print(f'Você escolheu jogar {num_rodadas} vezes.')

num_secreto = np.random.randint(1, 100 + 1)
print(num_secreto)

for i in range(num_rodadas):
    num_usuario = int(input("Agora escolha um número de 1 a 100!"))
    if num_usuario == num_secreto: print("Parabéns, você acertou!!")
    elif num_usuario < num_secreto: print("Não foi dessa vez, tente números maiores.")
    else: print("Não foi dessa vez, tente números menores.")
```
