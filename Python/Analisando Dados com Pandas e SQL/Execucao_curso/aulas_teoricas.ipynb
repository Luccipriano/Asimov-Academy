#### Obs.: sobre manipulação de índices em colunas ->

Por mais que não exista um comando "reset_columns", as colunas são índices assim como o que de fato é manipulado no reset_index.

É possível manipular os índices das colunas de duas formas:

##### Cenário 1: O "Truque da Transposição" (Literalmente resetar as colunas)
Se você quer pegar os nomes das suas colunas e transformá-los em uma linha de dados dentro do DataFrame (exatamente o que o reset_index faz com o índice das linhas), a maneira mais criativa e direta é transpor o DataFrame, resetar o índice e transpor de volta.

```python
import pandas as pd

"# Criando um DataFrame simples
df = pd.DataFrame({
    'Matemática': [90, 85],
    'História': [70, 75],
    'Física': [88, 92]
}, index=['Aluno A', 'Aluno B'])

print("--- Original ---")
print(df)

"# O TRUQUE:
"# 1. .T (Transpõe: colunas viram linhas)
"# 2. .reset_index() (O índice - antigas colunas - vira uma coluna de dados)
"# 3. .T (Transpõe de volta)
df_resetado = df.T.reset_index().T

print("\n--- Com Colunas 'Resetadas' para dentro dos dados ---")
print(df_resetado)
```

##### Cenário 2: Achatando Multi-Index (O caso mais comum)
Se você fez um groupby ou uma pivot table, provavelmente acabou com colunas hierárquicas (MultiIndex), onde você tem "níveis" de colunas (ex: ('Vendas', '2023'), ('Vendas', '2024')).

Nesse caso, o "reset" que a maioria das pessoas deseja é achatar esses níveis para ter apenas uma linha de cabeçalho simples.

```python
import pandas as pd

"# Criando um DataFrame com MultiIndex nas colunas (simulando um groupby)
cols = pd.MultiIndex.from_tuples([("Vendas", "Jan"), ("Vendas", "Fev"), ("Lucro", "Jan"), ("Lucro", "Fev")])
data = [[100, 120, 10, 12], [110, 130, 11, 13]]
df_multi = pd.DataFrame(data, columns=cols)

print("--- Original (MultiIndex) ---")
print(df_multi)

"# SOLUÇÃO: Achatar os níveis
"# Isso une os níveis com um sublinhado (ex: Vendas_Jan)
df_multi.columns = ['_'.join(col).strip() for col in df_multi.columns.values]

"# Se quiser apenas remover um nível específico (ex: ficar só com Jan/Fev e ignorar Vendas/Lucro):
"# df_multi.columns = df_multi.columns.get_level_values(1)

print("\n--- Achatado (Nível único) ---")
print(df_multi)
```

### Merge

````python
esquerda = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
'A': ['A0', 'A1', 'A2', 'A3'],
'B': ['B0', 'B1', 'B2', 'B3']})
esquerda
````

````python
direita = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
'C': ['C0', 'C1', 'C2', 'C3'],
'D': ['D0', 'D1', 'D2', 'D3']})
direita
````

````python
# É possível fazer um merge sem especificar a chave, de tal forma que o pandas vai automaticamente mesclar quanto existirem colunas com o nome igual. É análogo ao NATURAL JOIN em SQL
display(pd.merge(esquerda, direita))

# Também é possível especificar a coluna que vai ser usada no merge como chave, nesse caso redundante, mas nem sempre as colunas vão ter os mesmos nomes
# É comum vir indicando "id" quando for chave primária e "table_name_id" quando for chave estrangeira
display(pd.merge(esquerda, direita, on="key"))
````

````python
esquerda = pd.DataFrame({'key1': ['K0', 'K0', 'K1', 'K2'],
'key2': ['K0', 'K1', 'K0', 'K1'],
'A': ['A0', 'A1', 'A2', 'A3'],
'B': ['B0', 'B1', 'B2', 'B3']})
esquerda
````

````python
direita = pd.DataFrame({'key1': ['K0', 'K1', 'K1', 'K2'],
'key2': ['K0', 'K0', 'K0', 'K0'],
'C': ['C0', 'C1', 'C2', 'C3'],
'D': ['D0', 'D1', 'D2', 'D3']})
direita
````

````python
# Trabalhando com dados um pouco mais complexos, temos key1 e key2, permitindo executar merges com chaves mistas/compostas.
pd.merge(esquerda, direita, on=['key1', 'key2'])
````

````python
# Também é possível dizer qual o método de merge que vai ser usado
pd.merge(esquerda, direita, how='outer', on=['key1', 'key2'])
````

## Aplicando funções

É possível aplicar funções em todos os elementos de um dataframe ou de uma series, idêntico a aplicação de uma função em um vetor ou matriz.

````python
def comp(x):
    return x ** 2 + 3
````

````python
comp
## Identifica o ponteiro da memória onde a função está
````

````python
comp(2)
## Assim eu executo a função
````

````python
df['col2'].apply(comp) ## Usando o método apply, é possível aplicar uma função na lista de valores de uma series
````

````python
df['col1_calculada'] = df['col2'].apply(comp)
df['col1_calculada']

## Basicamente como colunas calculadas podem ser criadas com métodos de cálculos complexos que são abstraídos e depois executados de forma simples
## Excelente pra criação de métricas de KPIs
````

````python
df['col1'].apply(lambda x: x ** 2 + 3) 
# Também é possível aplicar funções lambda
````

É possível aplicar uma série de funções, cálculos e operações simples também.

````python
print("Soma:", df['col1'].sum())

print("Média:", df['col1'].mean())

print("Produto:", df['col1'].product())

print("Desvio Padrão:", df['col1'].std())

print("Máximo:", df['col1'].max())

print("Mínimo:", df['col1'].min())

## Esses dois são do caralho!!! Potencial enorme
print("Índice do valor máximo:", df['col1'].idxmax())
print("Índice do valor mínimo:", df['col1'].idxmin())

print("\n")
print("Soma Cumulativa:\n", df['col1'].cumsum())
````

````python
df[df["col2"] == 444]["col1"].sum()
# A combinação desses métodos gera possibilidades infinitas para resolução de problemas
````

````python
display(df.sort_values(by='col2'))
# Também é possível ordenar ou classificar o dataframe por qualquer coluna

display(df.sort_values(by=['col2', 'col1'], ascending=False))
# O filtro pode ser composto e misturado de forma ascendente ou descendente como bem entender
````
